```python
import matplotlib.pyplot as plt
import numpy as np
```

## Pie Charts ##
Circular chart divided into slices to show percentages of the total. Good for visualizing distribution among categories

```python
categories = ["Freshman", "Sophomores", "Juniors", "Seniors"]
values = np.array([300, 250, 275, 225])
color_values = ["red", "blue", "orange", "green"]

plt.pie(values, labels=categories,
        autopct="%1.1f%%",
        colors=color_values,
        explode=[0, 0, 0, 0.2],
        shadow=True,
        startangle=90)

plt.title("My College")



plt.show()
```
