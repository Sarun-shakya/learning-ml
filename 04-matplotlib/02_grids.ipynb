## grid() ##
Hepls make plots easier to read by adding reference lines

```python
x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]

plt.grid(axis="both",
         linewidth=2,
         color="lightgray",
         linestyle="dashed")

plt.plot(x, y)
plt.show()
```
