```python
import matplotlib.pyplot as plt
import numpy as np
```

## Bar Chart ##
Compare categories of data by representing each category

```python
categories = np.array(["Grains", "Fruit", "Vegetables", "Protein", "Dairy", "Sweets"])
values = np.array([4, 3, 1, 5, 3, 2])

plt.bar(categories, values, color="blue")
# plt.barh(categories, values, color="blue") # horizontal

plt.title("Daily Consumption")

plt.xlabel("Foods")
plt.ylabel("Quantity")

plt.show()
```
