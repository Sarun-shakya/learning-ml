```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([20, 30, 10, 15])

line_style = dict(marker=".",
         markersize=20, #ms
         markerfacecolor="cyan", #mfc
         markeredgecolor="red",
         linestyle="solid",
         linewidth=2,
         )

plt.title("Class Size", fontsize=20,
                        fontweight="bold",
                        color="purple")

plt.xlabel("Year",
           fontsize=16,
           fontweight="bold",
           color="blue")

plt.ylabel("Students",
           fontsize=16,
           fontweight="bold",
           color="blue")

plt.plot(x, y1, **line_style, color="blue")
plt.plot(x, y2, color="red",**line_style )

plt.xticks(x) # x at given number only
plt.tick_params(axis="both",
                colors="blue")
plt.show()
```
