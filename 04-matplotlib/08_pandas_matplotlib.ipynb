```python
df = pd.read_csv("data.csv")
print(df)
```

```python
print(df["Type1"].value_counts())
type_count = df["Type1"].value_counts(ascending=True)
```

```python
plt.barh(type_count.index, type_count.values, color="lightblue", edgecolor="black")
plt.title("No of Pokemon by Primary Type")
plt.xlabel("Count")
plt.ylabel("Type")
plt.tight_layout()
plt.show()
```
