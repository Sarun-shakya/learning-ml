## Filtering ##
Keeping the rows that match a condition

```python
print(df)
```

```python
tall_pokemon = df[df["Height"] >= 2.0]
print(tall_pokemon)
```

```python
heavy_pokemon = df[df["Weight"] > 100]
print(heavy_pokemon)
```

```python
legendary_pokemon = df[df["Legendary"] == 1]
print(legendary_pokemon)
```

```python
water_pokemon = df[df["Type1"] == "Water"]
print(water_pokemon)
```

```python
water_pokemon2 = df[(df["Type1"] == "Water") | (df["Type2"] == "Water")]
print(water_pokemon2)
```

```python
ff = df[(df["Type1"] == "Fire") & (df["Type2"] == "Flying")]
print(ff)
```
