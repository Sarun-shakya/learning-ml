## Series ##
1 dimensional labeled array that can hold any data type

```python
data = [100.1, 102.1, 104.1]
series = pd.Series(data)
print(series)
```

```python
data = ["A", "B", "C"]
series = pd.Series(data)
print(series)
```

```python
data = [True, False, True]
series = pd.Series(data)
print(series)
```

Custom label

```python
data = ["A", "B", "C"]
series = pd.Series(data, index = ["a", "b", "c"])
print(series)
```

```python
data = [1, 2, 3]
series = pd.Series(data, index=["a", "b", "c"])
print("Label a: " ,series.loc["a"])
series.loc["b"] = 200
print(series)
```

Locating with integer position

```python
data = [101, 102, 103]
series = pd.Series(data, index=["a", "b", "c"])
print(series.iloc[1])
```

Conditional locating

```python
print(series[series < 103])
```

```python
calories = {"Day 1": 1750, "Day 2": 2100, "Day 3": 1700}
# we dont need index here
series = pd.Series(calories)
print(series)
```

```python
print(series.loc["Day 3"])
series.loc["Day 3"] = 1800
print(series.loc["Day 3"])
```

Conditional Locating in dictionary

```python
print(series[series < 2000])
```

```python
data = {1: "Bulbasaur", 2: "Ivysaur", 3: "Venusaur", 4: "Charmander", 5: "Charmeleon", 6: "Charizard"}
series = pd.Series(data);
print(series)
```
