```python
import pandas as pd
```

## Data Frame ##
A tabular data structure with rows ans columns. (2 dimensional)

```python
data = {"Name": ["Ram", "Shyam", "Hari"],
        "Age": [20, 35, 20],
        }

df = pd.DataFrame(data, index=[1,2,3])
print(df)
```

```python
print(df.loc[2])
```

```python
print(df.iloc[0])
```

# Add a new Column #

```python
df["Job"] = ["Cook", "N/A", "Cashier"]
print(df)
```

# Add a new Row #

```python
new_row = pd.DataFrame([{"Name":"Sandy", "Age": 28, "Job": "nurse"}], index=[4])
df = pd.concat([df, new_row])
print(df)
```

# Add new rows #

```python
new_row = pd.DataFrame([{"Name":"Sandy", "Age": 28, "Job": "Nurse"},
                        {"Name":"Bandy", "Age": 60, "Job": "Manager"}],
                       index=[5,6])
df = pd.concat([df, new_row])
print(df)
```
