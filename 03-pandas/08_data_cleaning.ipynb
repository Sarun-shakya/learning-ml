## Data Cleaning ##
The process of fixing/removing: incomplete, incorrect, or irrelevent data. ~75% of work done with pandas is data cleaning

## Drop irrelevant columns ##

```python
df = df.drop(columns=["Legendary", "No"])
print(df)
```

## Handle missing data ##

```python
df = df.dropna(subset=["Type2"])
print(df.to_string())
```

```python
df = df.fillna({"Type2": "None"})
print(df.to_string())
```

## Fix inconsistent values ##

```python
df["Type1"] = df["Type1"].replace({"Grass": "GRASS", "Fire": "FIRE", "Water": "WATER"})
print(df.to_string)
```

## Stamdardize text ##

```python
df["Name"] = df["Name"].str.lower()
print(df)
```

## Fix Data Types ##

```python
df["Legendary"] = df["Legendary"].astype(bool)
print(df)
```

## Remove duplicate values ##

```python
df2 = pd.read_csv("data2.csv")
print(df2)
```

```python
df = df.drop_duplicates()
print(df)
```
